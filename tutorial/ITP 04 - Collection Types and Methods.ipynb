### In-place methods
Methods are functions that act on objects, and can either perform operations in-place (modify the underlying object it was called upon) or return a new object.
Notice that the method ```uper()```was not a stateful, in-place method as it returned a new string and did not modify the ```greeting``` variable. Take a look W3Schools provides information on other string methods in Python.

### Tab Completion
If you want to see a list of methods you can apply to an object, type `.` after the object, then hit tab key to see a drop down menu of available methods on that object.

Try it below on the `greeting` string object. Type `greeting`. then hit the Tab key.

````python
# Type . and hit Tab
gretting = "Hello"
gretting.
````

#### `help()`
While using tab completion is extremely helpful, if we use it to look through all possible methods ofr a given object, we might still not be certain how those methods work.

We can look up their documentation, or we can use the `help()`function we saw last lesson.

As a reminder, the `help()`funciton display some of the documentation for the item passed into it.

For example, when using tab completion above, we see the `capitalize()`string method, but we are not certain how it works.

````python
help(gretting.capitalize)
````
