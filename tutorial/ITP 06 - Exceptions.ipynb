### Exception Handling
Syntax Errors will always cause our programs to exit and fail, but we can hadle exceptions in Python. This allows us to program what happens in the case that Python encounters an exception or specific exceptions when trying to run a code block. To handle exceptions in Python we use a `try` statement.

We write `write` statement like this:

``` python
try:
    code block
except:
    code block
```
When Python encounters a `try` statement, it will first try to run the code in the `try` code block. If it encounters an exception, instead of existing and erroring like we have seen, it will instead run the code block under `except` .

````python
try:
    1 / 0 # Throws an exception
except:
    print("Exception Handled")
````

In this example, we run the `except` block if we encounter any exception in the `try` block. If we want to only handle a certain exceptioni we can write the exception after the `except` keyword like this:

``` python
try:
    code block:
except ExceptionName:
    code block
```

````python
try:
    1 / 0 # Throws a ZeroDivisionError
except ZeroDivisionError:
    print("Exception Handled")
````

If we want to handle multiple specific exceptions we can write a sequence of exceptions separated by commas inside parentheses.

Try commenting out one of the exception throwing lines below at a time, and notice that both exceptions are handled.

````python
try:
    #\1 / 0 # Throws a ZeroDivisionError
    print(undefined_variable) # Throws a NameError exception
except (ZeroDivisionError, NameError):
    print("Exception Handled")
````

This now handles both ZeroDivisionError and NameError exceptions.

#### Assertion Error

One very useful exception is called an `AssertionError`. We can rise `AssertionErrors` using `assert` statements. We have been using these to check your lab code during this course. We use an `assert` statement like this: 

`assert boolean expression, option message`

When Python executes this statement, it evaluates the boolean expression first. If its `true`, Python does nothings and moves on. If it is `false`, then it throws an AssertionError with the optional message, if provided.

````python
assert 1 == 1
assert 1 == 2, "That is no true"
````
