###  **B. Reading Data**

#### **1. Schema Inference**

Let's start by creating a DataFrame by reading a directory of CSV files.

```python
# Read the customers.csv file with schema inference and print the schema
from pyspark.sql import SparkSession

# Create a the first SparkSession
spark = SparkSession.builder.appName("FirstSession").getOrCreate()

# Read the CSV file with header and inferSchema options
customer_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("../resources/customers.csv")
```

```python
customer_df.printSchema()
```

```python
# Alternatively we can use below command to print schema in linear format:
print(customer_df.schema)
```

####  **`display()` function**

The `display()` function is available in Databrics runtimes to be used within notebooks.

`display()` represents a Spark action, which returns results for a DataFrame in a formatted tabular result window. As this is intented for visual inspection only, results are limited. Additional fetaures of the `display()` function include:
* Ability to sort or filter data in the result set.
* Ability to download result data to a CSV file
* Ability to profile data usgin the + -> *Data Profile* option
* Ability to visualize data using the + -> *Visualization* option

```python
# Using Display function to preview the data in the DataFrame
display(customer_df)
```
