###  **B. Reading Data**

#### **1. Schema Inference**

Let's start by creating a DataFrame by reading a directory of CSV files.

```python
# Read the customers.csv file with schema inference and print the schema
from pyspark.sql import SparkSession

# Create a the first SparkSession
spark = SparkSession.builder.appName("FirstSession").getOrCreate()

# Read the CSV file with header and inferSchema options
customer_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("../resources/customers.csv")
```

```python
customer_df.printSchema()
```

```python
# Alternatively we can use below command to print schema in linear format:
print(customer_df.schema)
```

####  **`display()` function**

The `display()` function is available in Databrics runtimes to be used within notebooks.

`display()` represents a Spark action, which returns results for a DataFrame in a formatted tabular result window. As this is intented for visual inspection only, results are limited. Additional fetaures of the `display()` function include:
* Ability to sort or filter data in the result set.
* Ability to download result data to a CSV file
* Ability to profile data usgin the + -> *Data Profile* option
* Ability to visualize data using the + -> *Visualization* option

```python
# Using Display function to preview the data in the DataFrame
display(customer_df)
```

#### **3. Using a DDL Schema**

Let's define the schema now using a DDL schema for readability. 

```python
ddl_schema = """customer_id INT, 
    first_name STRING, 
    last_name STRING, 
    age INT, 
    email STRING, 
    join_date STRING, 
    last_login STRING, 
    total_purchases DOUBLE
"""

customers_ddl_df =  spark.read.format("csv") \
    .option("header", "true") \
    .schema(ddl_schema) \
    .load("../resources/customers.csv")
```

```python
customers_ddl_df.printSchema()
```

###  **C. Writing Data From DataFrames**

Now let's write out the contents of a DataFrame to different output locations.

#### **Writing the contents of a DataFrame to a File System**

We can write the contents of our DataFrame to a filesystem in various formats using the `write`and `save`methods as shown here.

```python
# Define an output path
parquet_output_volume_path = "../output/parquet/customers/"
```

```python
# Write the DataFrame out as Parquet files to a directory
customers_ddl_df.write.format("parquet") \
    .mode("overwrite") \
    .save(parquet_output_volume_path)
```

```python
# Show the parquet files's data written to the output path
import os
os.listdir(parquet_output_volume_path)
```

#### **Writing the contents of a DataFrame to a Table**

We can also save our DataFrame to a table (defined in a catalog and schema in Unity Catalo). 

```python
# Writing our DataFrame in a new table (this is an action)
customers_ddl_df.write.saveAsTable("customers_table", mode="overwrite")
```
